```python
import pandas as pd
import plotly.express as px

# 1. Load the cleaned data
print("Loading Cleaned Data...")
df = pd.read_csv('cleaned_agri_data.csv')

# Let's ensure our column names are exact based on the cleaning step
# (They should be lowercase with underscores, e.g., 'state_name', 'rice_production_1000_tons')

# ==========================================
# CHART 1: Top 7 RICE PRODUCTION State Data (Bar Plot)
# ==========================================
# Group data by State and sum the rice production
rice_state = df.groupby('state_name')['rice_production_1000_tons'].sum().reset_index()
# Sort from highest to lowest and grab the top 7
top_7_rice = rice_state.sort_values(by='rice_production_1000_tons', ascending=False).head(7)

fig1 = px.bar(top_7_rice, x='state_name', y='rice_production_1000_tons', 
              title='Top 7 States for Rice Production',
              labels={'state_name': 'State', 'rice_production_1000_tons': 'Production (1000 Tons)'},
              color='rice_production_1000_tons', color_continuous_scale='Greens')
fig1.show()

# ==========================================
# CHART 2 & 3: Top 5 Wheat Producing States (Bar & Pie Chart)
# ==========================================
wheat_state = df.groupby('state_name')['wheat_production_1000_tons'].sum().reset_index()
top_5_wheat = wheat_state.sort_values(by='wheat_production_1000_tons', ascending=False).head(5)

# Bar Chart
fig2 = px.bar(top_5_wheat, x='state_name', y='wheat_production_1000_tons',
              title='Top 5 States for Wheat Production',
              labels={'state_name': 'State', 'wheat_production_1000_tons': 'Production (1000 Tons)'},
              color='wheat_production_1000_tons', color_continuous_scale='Oranges')
fig2.show()

# Pie Chart for Percentage
fig3 = px.pie(top_5_wheat, names='state_name', values='wheat_production_1000_tons',
              title='Top 5 Wheat Producing States (Percentage Contribution)',
              hole=0.3) # Makes it a donut-style pie chart
fig3.show()

# ==========================================
# CHART 4: India's SUGARCANE PRODUCTION Over Time (Line Plot)
# ==========================================
# Group data by Year and sum the sugarcane production across all India
sugar_year = df.groupby('year')['sugarcane_production_1000_tons'].sum().reset_index()

fig4 = px.line(sugar_year, x='year', y='sugarcane_production_1000_tons',
               title="India's Sugarcane Production Trend",
               labels={'year': 'Year', 'sugarcane_production_1000_tons': 'Total Production (1000 Tons)'},
               markers=True)
fig4.update_traces(line_color='purple')
fig4.show()

print("Batch 1 Charts Generated Successfully! Hover over the bars and lines to see exact numbers.")
```

```python
# ==========================================
# CHART 5: Oil Seed Production by Top 5 States
# ==========================================
oil_state = df.groupby('state_name')['oilseeds_production_1000_tons'].sum().reset_index()
top_5_oil = oil_state.sort_values(by='oilseeds_production_1000_tons', ascending=False).head(5)

fig5 = px.bar(top_5_oil, x='state_name', y='oilseeds_production_1000_tons',
              title='Top 5 States for Oilseed Production',
              labels={'state_name': 'State', 'oilseeds_production_1000_tons': 'Production (1000 Tons)'},
              color='oilseeds_production_1000_tons', color_continuous_scale='YlOrBr')
fig5.show()

# ==========================================
# CHART 6: Top 7 Sunflower Production States
# ==========================================
sun_state = df.groupby('state_name')['sunflower_production_1000_tons'].sum().reset_index()
top_7_sun = sun_state.sort_values(by='sunflower_production_1000_tons', ascending=False).head(7)

fig6 = px.bar(top_7_sun, x='state_name', y='sunflower_production_1000_tons',
              title='Top 7 States for Sunflower Production',
              labels={'state_name': 'State', 'sunflower_production_1000_tons': 'Production'},
              color='sunflower_production_1000_tons', color_continuous_scale='Sunset')
fig6.show()

# ==========================================
# CHART 7: Rice vs Wheat Production Over Time (Line Plot)
# ==========================================
# Group data by year for both crops
rw_year = df.groupby('year')[['rice_production_1000_tons', 'wheat_production_1000_tons']].sum().reset_index()

fig7 = px.line(rw_year, x='year', y=['rice_production_1000_tons', 'wheat_production_1000_tons'],
               title='Rice vs Wheat Production in India Over Time',
               labels={'value': 'Total Production (1000 Tons)', 'year': 'Year', 'variable': 'Crop Type'},
               markers=True)
fig7.show()

# ==========================================
# CHART 8: Rice Production by West Bengal Districts
# ==========================================
# Filter data for only West Bengal
wb_data = df[df['state_name'].str.contains('West Bengal', case=False, na=False)]
wb_dist = wb_data.groupby('dist_name')['rice_production_1000_tons'].sum().reset_index()
# Getting the top 15 districts so the chart isn't too crowded
top_wb_dist = wb_dist.sort_values(by='rice_production_1000_tons', ascending=False).head(15)

fig8 = px.bar(top_wb_dist, x='dist_name', y='rice_production_1000_tons',
              title='Top 15 Rice Producing Districts in West Bengal',
              labels={'dist_name': 'District', 'rice_production_1000_tons': 'Production'},
              color='rice_production_1000_tons', color_continuous_scale='Teal')
fig8.show()

# ==========================================
# CHART 9: Top 10 Wheat Production Years in UP
# ==========================================
# Filter data for only Uttar Pradesh
up_data = df[df['state_name'].str.contains('Uttar Pradesh', case=False, na=False)]
up_years = up_data.groupby('year')['wheat_production_1000_tons'].sum().reset_index()
top_10_up_years = up_years.sort_values(by='wheat_production_1000_tons', ascending=False).head(10)

# Convert year to string so the bar chart treats it as a category, not a continuous number
top_10_up_years['year'] = top_10_up_years['year'].astype(str)

fig9 = px.bar(top_10_up_years, x='year', y='wheat_production_1000_tons',
              title='Top 10 Best Years for Wheat Production in Uttar Pradesh',
              labels={'year': 'Year', 'wheat_production_1000_tons': 'Production'},
              color='wheat_production_1000_tons', color_continuous_scale='Burg')
fig9.show()

print("Batch 2 Charts Generated Successfully!")
```

```python
import plotly.graph_objects as go
import plotly.express as px

# We are using the exact spelling from the dataset to make it 100% crash-proof
soy_prod_col = 'soyabean_production_1000_tons'
soy_yield_col = 'soyabean_yield_kg_per_ha'

# ==========================================
# CHART 12: Soybean Production & Yield (Top 5 States)
# ==========================================
soy_state = df.groupby('state_name').agg({
    soy_prod_col: 'sum',
    soy_yield_col: 'mean' # Average yield
}).reset_index()

top_5_soy = soy_state.sort_values(by=soy_prod_col, ascending=False).head(5)

# Using a scatter plot where the size of the bubble is the Production, and Y-axis is Yield
fig12 = px.scatter(top_5_soy, x='state_name', y=soy_yield_col, size=soy_prod_col,
                 color='state_name', title='Soyabean: Yield Efficiency vs Total Production (Top 5 States)',
                 labels={soy_yield_col: 'Average Yield (Kg/ha)', 'state_name': 'State'},
                 size_max=40)
fig12.show()

# ==========================================
# CHART 13: Impact of Area Cultivated on Production (Scatter Plot)
# ==========================================
# We will filter out zero values to make the chart readable
area_prod_df = df[(df['rice_area_1000_ha'] > 0) & (df['rice_production_1000_tons'] > 0)]

fig13 = px.scatter(area_prod_df, x='rice_area_1000_ha', y='rice_production_1000_tons', 
                 opacity=0.5, color='state_name',
                 title='Impact of Area Cultivated on Rice Production',
                 labels={'rice_area_1000_ha': 'Area Cultivated (1000 ha)', 'rice_production_1000_tons': 'Production (1000 tons)'})
fig13.show()

# ==========================================
# CHART 14: Rice vs. Wheat Yield Across Top States
# ==========================================
yield_state = df.groupby('state_name')[['rice_yield_kg_per_ha', 'wheat_yield_kg_per_ha']].mean().reset_index()
# Grab top states based on combined yield just to keep the chart clean
yield_state['total_yield'] = yield_state['rice_yield_kg_per_ha'] + yield_state['wheat_yield_kg_per_ha']
top_yield_states = yield_state.sort_values(by='total_yield', ascending=False).head(10)

fig14 = go.Figure(data=[
    go.Bar(name='Rice Yield', x=top_yield_states['state_name'], y=top_yield_states['rice_yield_kg_per_ha']),
    go.Bar(name='Wheat Yield', x=top_yield_states['state_name'], y=top_yield_states['wheat_yield_kg_per_ha'])
])
fig14.update_layout(title='Average Rice vs Wheat Yield Across Top 10 States', barmode='group')
fig14.show()

print("✅ Phase 2 (EDA) is 100% Complete! You have generated all required Python Visualizations.")
```
