```python
import sqlite3
import pandas as pd

# 1. Load the cleaned data
df = pd.read_csv('cleaned_agri_data.csv')

# 2. Create and connect to a local SQL Database
print("🔌 Connecting to SQL Database...")
conn = sqlite3.connect('agridata.db')

# 3. Load the data into a SQL table named 'crop_data'
df.to_sql('crop_data', conn, if_exists='replace', index=False)
print("✅ Database created and data loaded into 'crop_data' table!\n")

# Function to run SQL queries and display them nicely
def run_query(query_title, sql_query):
    print(f"--- {query_title} ---")
    result = pd.read_sql_query(sql_query, conn)
    display(result)
    print("\n")

# ==========================================
# SQL QUESTION 1: Year-wise Trend of Rice Production Across States (Top 3)
# ==========================================
q1 = """
SELECT state_name, year, SUM(rice_production_1000_tons) as total_rice_production
FROM crop_data
WHERE state_name IN (
    SELECT state_name 
    FROM crop_data 
    GROUP BY state_name 
    ORDER BY SUM(rice_production_1000_tons) DESC 
    LIMIT 3
)
GROUP BY state_name, year
ORDER BY state_name, year;
"""
run_query("Q1: Year-wise Trend of Rice Production (Top 3 States)", q1)

# ==========================================
# SQL QUESTION 2: Top 5 Districts by Wheat Yield (Highest ever recorded)
# (Adjusted slightly to ensure it works across all dataset years)
# ==========================================
q2 = """
SELECT state_name, dist_name, MAX(wheat_yield_kg_per_ha) as max_wheat_yield
FROM crop_data
WHERE wheat_yield_kg_per_ha > 0
GROUP BY state_name, dist_name
ORDER BY max_wheat_yield DESC
LIMIT 5;
"""
run_query("Q2: Top 5 Districts by Highest Wheat Yield", q2)

# ==========================================
# SQL QUESTION 3: States with the Highest Total Oilseed Production
# ==========================================
q3 = """
SELECT state_name, SUM(oilseeds_production_1000_tons) as total_oilseed_production
FROM crop_data
GROUP BY state_name
ORDER BY total_oilseed_production DESC
LIMIT 5;
"""
run_query("Q3: States with Highest Oilseed Production", q3)

# ==========================================
# SQL QUESTION 4: District-wise Area and Production for Major Crops
# ==========================================
q4 = """
SELECT dist_name, state_name, 
       SUM(rice_area_1000_ha) as total_rice_area, SUM(rice_production_1000_tons) as total_rice_prod,
       SUM(wheat_area_1000_ha) as total_wheat_area, SUM(wheat_production_1000_tons) as total_wheat_prod
FROM crop_data
GROUP BY dist_name, state_name
ORDER BY total_rice_prod DESC
LIMIT 5;
"""
run_query("Q4: District-wise Area & Production for Rice and Wheat (Top 5 Rice Districts)", q4)

# ==========================================
# SQL QUESTION 5: Yearly Production Growth of Cotton in Top 5 States
# ==========================================
q5 = """
SELECT state_name, year, SUM(cotton_production_1000_tons) as total_cotton
FROM crop_data
WHERE state_name IN (
    SELECT state_name 
    FROM crop_data 
    GROUP BY state_name 
    ORDER BY SUM(cotton_production_1000_tons) DESC 
    LIMIT 5
)
GROUP BY state_name, year
ORDER BY state_name, year;
"""
run_query("Q5: Yearly Cotton Production in Top 5 States", q5)
```

```python
# ==========================================
# SQL QUESTION 6: Districts with the Highest Groundnut Production in 2017
# ==========================================
q6 = """
SELECT dist_name, state_name, groundnut_production_1000_tons
FROM crop_data
WHERE year = 2017
ORDER BY groundnut_production_1000_tons DESC
LIMIT 5;
"""
run_query("Q6: Highest Groundnut Production in 2017", q6)

# ==========================================
# SQL QUESTION 7: Annual Average Maize Yield Across All States
# ==========================================
q7 = """
SELECT year, AVG(maize_yield_kg_per_ha) as avg_maize_yield
FROM crop_data
WHERE maize_yield_kg_per_ha > 0
GROUP BY year
ORDER BY year DESC
LIMIT 10;
"""
run_query("Q7: Annual Average Maize Yield (Last 10 Recorded Years)", q7)

# ==========================================
# SQL QUESTION 8: Total Area Cultivated for Oilseeds in Each State
# ==========================================
q8 = """
SELECT state_name, SUM(oilseeds_area_1000_ha) as total_oilseed_area
FROM crop_data
GROUP BY state_name
ORDER BY total_oilseed_area DESC
LIMIT 10;
"""
run_query("Q8: Total Area Cultivated for Oilseeds (Top 10 States)", q8)

# ==========================================
# SQL QUESTION 9: Districts with the Highest Rice Yield
# ==========================================
q9 = """
SELECT dist_name, state_name, MAX(rice_yield_kg_per_ha) as max_rice_yield
FROM crop_data
WHERE rice_yield_kg_per_ha > 0
GROUP BY dist_name, state_name
ORDER BY max_rice_yield DESC
LIMIT 5;
"""
run_query("Q9: Districts with the Highest Rice Yield", q9)

# ==========================================
# SQL QUESTION 10: Compare Wheat and Rice Production for Top 5 States
# ==========================================
q10 = """
SELECT state_name, 
       SUM(wheat_production_1000_tons) as total_wheat, 
       SUM(rice_production_1000_tons) as total_rice
FROM crop_data
GROUP BY state_name
ORDER BY (total_wheat + total_rice) DESC
LIMIT 5;
"""
run_query("Q10: Compare Wheat vs Rice Production in Top 5 Overall States", q10)

print("✅ Phase 3 (SQL Analysis) is 100% Complete!")
```
