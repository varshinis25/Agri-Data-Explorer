```python
import pandas as pd
import numpy as np

# 1. Load the dataset
print("Loading Agricultural Data...")
# Change the filename below if yours is different (e.g., 'agri_data.xlsx')
try:
    df = pd.read_csv('agri_data.csv') 
except FileNotFoundError:
    print("Error: Could not find 'agri_data.csv'. Please ensure the file is in the folder and named correctly.")
    df = None

if df is not None:
    # 2. Basic Inspection
    print(f"\nDataset Shape: {df.shape} rows and {df.shape} columns")
    print("\nFirst 3 rows:")
    display(df.head(3))

    # 3. Clean Column Names (Standardize them by making them lowercase and replacing spaces with underscores)
    print("\nCleaning column names...")
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
    
    # Let's see the new column names
    print(df.columns.tolist()[:10]) # Print first 10 just to check

    # 4. Handle Missing Values
    print("\nChecking for missing values...")
    missing_data = df.isnull().sum()
    print(missing_data[missing_data > 0]) # Only print columns that actually have missing values

    # Strategy for missing agricultural data: 
    # If a yield or production is missing, it often means that crop wasn't grown there that year.
    # We will fill missing numeric values with 0. 
    # NOTE: In a real-world scenario, you might want to interpolate or leave them as NaN depending on the specific analysis, but filling with 0 is safest for basic SQL/Power BI aggregations without breaking things.
    
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    df[numeric_cols] = df[numeric_cols].fillna(0)

    # 5. Save the Cleaned Data
    # We will save this as a new CSV to use for EDA and SQL later.
    clean_filename = 'cleaned_agri_data.csv'
    df.to_csv(clean_filename, index=False)
    print(f"\n✅ Data cleaning complete. Cleaned data saved to '{clean_filename}'")
```
